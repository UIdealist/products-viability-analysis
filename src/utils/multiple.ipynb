```python
import sys
from pathlib import Path
def find_src_folder(current_path: Path, folder_name: str = 'src') -> Path:
    search_directories = [current_path] + list(current_path.parents)
    for parent in search_directories:
        if parent.name == folder_name:
            return parent.parent
    return current_path
src_path = find_src_folder(Path.cwd(), 'src')
sys.path.append(str(src_path))
```

```python
import matplotlib.pyplot as plt
import numpy as np

industries = [
    "Redes Sociales",
    "Software",
    "Servicios",
    "Entretenimiento",
    "Comercio Electrónico",
    "Música/Audio",
    "Foto/Video",
    "Analítica/Big Data",
    "Móvil",
    "Desarrollo Web"
]

values = [12.3, 9.3, 8.3, 6.9, 5.9, 4.4, 4.4, 3.9, 3.9, 3.9]

plt.rcParams["font.family"] = "Times New Roman"
fig, ax = plt.subplots(figsize=(10, 5))

y = np.arange(len(industries))

bars = ax.barh(
    y,
    values,
    color="#687B8B",
    edgecolor="white",
    alpha=0.85
)

for i, v in enumerate(values):
    ax.text(
        v + 0.15,
        i,
        f"{v:.2f} %",
        va="center",
        ha="left",
        fontsize=9,
        color="white"
    )

ax.set_yticks(y)
ax.set_yticklabels(industries, fontsize=10)
ax.invert_yaxis()

ax.set_xlim(0, 13)
ax.set_xticks(np.arange(0, 13, 2))
ax.set_xticklabels([f"{i} %" for i in range(0, 13, 2)], fontsize=9)

ax.xaxis.grid(True, linestyle=":", linewidth=1, color="#DCDCDC")
ax.yaxis.grid(False)
ax.set_axisbelow(True)

ax.set_title("Industria", fontsize=14, pad=12)

for spine in ["top", "right", "left", "bottom"]:
    ax.spines[spine].set_visible(False)

ax.tick_params(axis="x", length=0)
ax.tick_params(axis="y", length=0)

plt.tight_layout()
plt.show()
```
